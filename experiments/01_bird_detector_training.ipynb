BirdDetectopr training using dataset_v1 (train - yalour_green, validation - moot)

```python
from deepforest import main
import pandas as pd
import matplotlib.pyplot as plt
import cv2
from pathlib import Path
from pytorch_lightning.callbacks import Callback
import os
import torch
from pytorch_lightning.callbacks import ModelCheckpoint
from deepforest import evaluate
import pandas as pd
```

```python
config = {
    "model": {
        "num_classes": 1,
        "label_dict": {"Penguine": 0}
    }
}
```

```python
model = main.deepforest(config_args=config)
model.load_model(model_name="weecology/deepforest-bird")
model.label_dict = {"Penguine": 0}
model.config["train"]["epochs"] = 400
model.config["train"]["batch_size"] = 4
model.config["train"]["lr"] = 1e-4

model.config["train"]["csv_file"] = "data/train.csv"
model.config["train"]["root_dir"] = "data/cut_yalur_green_v2/images"

model.config["validation"]["csv_file"] = "data/valid.csv"
model.config["validation"]["root_dir"] = "data/cut_moot_100m/images"
model.config["validation"]["iou_threshold"] = 0.5
model.config["train"]["max_detection_threshold"] = 1000
model.config["validation"]["max_detection_threshold"] = 1000
```

```python
print(model.label_dict)
```

```python
checkpoint = ModelCheckpoint(
    monitor="val_loss",
    mode="min",
    save_top_k=3,
    save_last=True,
    dirpath="diploma_v3/runs/",
    filename="deepforest-{epoch:02d}-{val_loss:.3f}"
)
```

```python
model.create_trainer(callbacks=[checkpoint])

model.trainer.fit(model)
```

```python
best_model_path = checkpoint.best_model_path
print("BEST MODEL:", best_model_path)
```
