```python
from pathlib import Path
import cv2
```

```python
def analyze_penguin_distribution(labels_path, images_path, verbose=True):

    labels_path = Path(labels_path)
    images_path = Path(images_path)

    total_images = 0
    images_with_penguins = 0
    images_without_penguins = 0
    total_penguins = 0

    total_bbox_area_px = 0
    total_bbox_width_px = 0
    total_bbox_height_px = 0
    total_bboxes = 0

    image_extensions = [".jpg", ".jpeg", ".png", ".JPG", ".PNG"]

    for label_file in labels_path.glob("*.txt"):
        total_images += 1

        image_file = None
        for ext in image_extensions:
            candidate = images_path / f"{label_file.stem}{ext}"
            if candidate.exists():
                image_file = candidate
                break

        if image_file is None:
            print(f"Image not found for: {label_file.stem}")
            continue

        img = cv2.imread(str(image_file))
        if img is None:
            print(f"Failed to read image: {image_file}")
            continue

        img_h, img_w = img.shape[:2]

        with open(label_file, "r") as f:
            lines = [line.strip() for line in f.readlines() if line.strip()]

        if len(lines) > 0:
            images_with_penguins += 1
            total_penguins += len(lines)

            for line in lines:
                parts = line.split()
                if len(parts) == 5:
                    _, _, _, w, h = map(float, parts)

                    w_px = w * img_w
                    h_px = h * img_h
                    area_px = w_px * h_px

                    total_bbox_width_px += w_px
                    total_bbox_height_px += h_px
                    total_bbox_area_px += area_px
                    total_bboxes += 1
        else:
            images_without_penguins += 1

    empty_percentage = (images_without_penguins / total_images * 100) if total_images else 0

    avg_penguins_per_non_empty_image = (
        total_penguins / images_with_penguins if images_with_penguins else 0
    )

    avg_width_px = total_bbox_width_px / total_bboxes if total_bboxes else 0
    avg_height_px = total_bbox_height_px / total_bboxes if total_bboxes else 0
    avg_area_px = total_bbox_area_px / total_bboxes if total_bboxes else 0

    results = {
        "total_images": total_images,
        "empty_percentage": empty_percentage,
        "total_penguins": total_penguins,
        "avg_penguins_per_non_empty_image": avg_penguins_per_non_empty_image,
        "avg_width_px": avg_width_px,
        "avg_height_px": avg_height_px,
        "avg_area_px": avg_area_px
    }

    if verbose:
        print(f"\nLabels: {labels_path}")
        print(f"Images: {images_path}")
        print(f"Total images: {total_images}")
        print(f"Empty images: {empty_percentage:.2f}%")
        print(f"Total penguins: {total_penguins}")
        print(f"Avg penguins (non-empty images): {avg_penguins_per_non_empty_image:.2f}")
        print(f"Avg bbox size: {avg_width_px:.1f}px × {avg_height_px:.1f}px")
        print(f"Avg bbox area: {avg_area_px:.1f} px²")

    return results
```

```python
def combine_dataset_stats(*datasets, verbose=True):
    """
    Combines multiple dataset stats and computes overall empty image percentage.
    
    Args:
        *datasets: dictionaries returned by analyze_penguin_distribution
    """

    total_images = 0
    total_empty_images = 0

    for ds in datasets:
        ds_total = ds["total_images"]
        ds_empty = ds_total * (ds["empty_percentage"] / 100)

        total_images += ds_total
        total_empty_images += ds_empty

    combined_empty_percentage = (
        total_empty_images / total_images * 100 if total_images else 0
    )

    if verbose:
        print(f"Total images: {total_images}")
        print(f"Empty images: {combined_empty_percentage:.2f}%")

    return {
        "total_images": total_images,
        "empty_percentage": combined_empty_percentage
    }
```
