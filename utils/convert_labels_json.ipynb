```python
from pathlib import Path
import json
from PIL import Image

images_dir = Path("data/cut_test/images/test")
labels_dir = Path("data/cut_test/labels/test")
 
tasks = []

for img_path in images_dir.glob("*.jpg"):

    label_path = labels_dir / f"{img_path.stem}.txt"

    predictions = []

    if label_path.exists():
        for line in label_path.read_text().splitlines():

            cls, x, y, w, h = map(float, line.split()[:5])

            predictions.append({
                "from_name": "label",
                "to_name": "image",
                "type": "rectanglelabels",
                "value": {
                    "x": (x - w/2) * 100,
                    "y": (y - h/2) * 100,
                    "width": w * 100,
                    "height": h * 100,
                    "rectanglelabels": ["Penguine"]
                }
            })

    tasks.append({
        "data": {
            "image": f"/data/local-files/?d=data/cut_test/images/test/{img_path.name}"
        },
        "predictions": [{
            "result": predictions
        }]
    })

with open("label_studio_cut_images_test.json", "w") as f:
    json.dump(tasks, f)
```
