```python
import os
import shutil

source_images = "cut_unlabeled/images"
labels_folder = "labeled_unlabeled_dataset/labels"
target_images = "labeled_unlabeled_dataset/images"

os.makedirs(target_images, exist_ok=True)

label_basenames = {os.path.splitext(f)[0] for f in os.listdir(labels_folder) if f.endswith(".txt")}

copied = 0

for img_file in os.listdir(source_images):
    img_path = os.path.join(source_images, img_file)

    if not os.path.isfile(img_path):
        continue

    base_name = os.path.splitext(img_file)[0]

    if base_name in label_basenames:
        dst_path = os.path.join(target_images, img_file)
        shutil.copy2(img_path, dst_path)
        copied += 1

print(f"Copied {copied} images with labels to {target_images}")
```
