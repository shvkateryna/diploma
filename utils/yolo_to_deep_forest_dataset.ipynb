```python
import os
import pandas as pd
from pathlib import Path
import cv2

images_dir = "data/cut_yalur_green_v2/images"
labels_dir = "data/cut_yalur_green_v2/labels"
output_csv = "train.csv"

label_name = "Penguine"

rows = []

image_paths = list(Path(images_dir).glob("*.*"))

for img_path in image_paths:

    label_path = Path(labels_dir) / (img_path.stem + ".txt")

    if not label_path.exists():

        rows.append([
            img_path.name,
            0, 0, 0, 0,
            label_name
        ])

        continue

    img = cv2.imread(str(img_path))
    h, w = img.shape[:2]

    with open(label_path) as f:
        lines = f.readlines()

    if len(lines) == 0:

        rows.append([
            img_path.name,
            0, 0, 0, 0,
            label_name
        ])

        continue

    for line in lines:

        cls, xc, yc, bw, bh = map(float, line.split())

        xmin = int((xc - bw/2) * w)
        ymin = int((yc - bh/2) * h)
        xmax = int((xc + bw/2) * w)
        ymax = int((yc + bh/2) * h)

        rows.append([
            img_path.name,
            xmin,
            ymin,
            xmax,
            ymax,
            label_name
        ])

df = pd.DataFrame(
    rows,
    columns=["image_path","xmin","ymin","xmax","ymax","label"]
)

df.to_csv(output_csv, index=False)

print("saved:", output_csv)
print("rows:", len(df))
print("images:", len(image_paths))
```
